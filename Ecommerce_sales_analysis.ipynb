```python
total_sales = df["Sales"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Profit"].sum()

print("Total Sales:", total_sales)
print("Total Cost:", total_cost)
print("Total Profit:", total_profit)
```

```python
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum().sort_values("Sales", ascending=False)

category_analysis
```

```python
category_analysis["Sales"].plot(kind="bar", figsize=(8,5))

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_analysis["Profit"].plot(kind="bar", figsize=(8,5))

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum().sort_values("Sales", ascending=False)

region_analysis
```

```python
region_analysis["Sales"].plot(kind="bar", figsize=(8,5))

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_analysis["Profit"].plot(kind="bar", figsize=(8,5))

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    df.groupby(df["Order_Date"].dt.month)["Sales"]
    .sum()
)

monthly_sales
```

```python
monthly_sales.index = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

monthly_sales.plot(kind="line", marker="o", figsize=(10,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_products_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_sales
```

```python
top_products_sales.sort_values().plot(
    kind="barh",
    figsize=(9,6)
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
top_products_profit = (
    df.groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_profit
```

```python
top_products_profit.sort_values().plot(
    kind="barh",
    figsize=(9,6)
)

plt.title("Top 10 Products by Profit")
plt.xlabel("Total Profit")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
payment_analysis = (
    df.groupby("Payment_Mode")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

payment_analysis
```

```python
payment_analysis.plot(
    kind="bar",
    figsize=(9,5)
)

plt.title("Sales by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
segment_analysis = (
    df.groupby("Customer_Segment")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

segment_analysis
```

```python
segment_analysis["Sales"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
segment_analysis["Profit"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_margin = (
    df.groupby("Category")[["Sales", "Profit"]]
    .sum()
)

category_margin["Profit_Margin_%"] = (
    category_margin["Profit"] / category_margin["Sales"] * 100
).round(2)

category_margin.sort_values(
    "Profit_Margin_%",
    ascending=False
)
```

```python
category_margin["Profit_Margin_%"].sort_values().plot(
    kind="barh",
    figsize=(9,5)
)

plt.title("Profit Margin by Category")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

```python
loss_orders = df[df["Profit"] < 0]

print("Loss-Making Orders:", len(loss_orders))
print("Total Loss:", loss_orders["Profit"].sum())
```

```python
loss_by_category = (
    loss_orders.groupby("Category")["Profit"]
    .sum()
    .sort_values()
)

loss_by_category
```

```python
loss_by_product = (
    loss_orders.groupby("Product")["Profit"]
    .sum()
    .sort_values()
)

loss_by_product
```

```python
loss_by_product.plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Loss by Product")
plt.xlabel("Product")
plt.ylabel("Total Loss")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
overall_profit_margin = (
    total_profit / total_sales * 100
)

print("Overall Profit Margin:", round(overall_profit_margin, 2), "%")
```
